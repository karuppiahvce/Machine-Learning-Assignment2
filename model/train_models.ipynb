```python
# ELECTRICAL GRID STABILITY CLASSIFICATION PROBLEM

# STEP 1: Import libraries

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report
)

print("Libraries imported successfully!")

# LOAD THE ORIGINAL DATASET

data = pd.read_csv("Grid stability test dataset.csv")

print("\nOriginal Dataset")
print("----------------")
print("Number of rows    :", data.shape[0])
print("Number of columns :", data.shape[1])

print("\nFirst 5 rows:")
display(data.head())

# CHECK THE DATASET

print("\nColumn names:")
print(data.columns.tolist())

print("\nMissing values:")
print(data.isnull().sum())

print("\nTarget distribution:")
print(data["stabf"].value_counts())

# CREATE TRAINING AND TEST DATA

# 80% Training data
# 20% Test data

train_data, test_data = train_test_split(
    data,
    test_size=0.20,
    random_state=42,
    stratify=data["stabf"]
)

print("\nTraining data size:", train_data.shape)
print("Test data size    :", test_data.shape)

print("TRAINING DATASET - First 10 Rows")
display(train_data.head(10))

print("TEST DATASET - First 10 Rows")
display(test_data.head(10))

# Save the two datasets

train_data.to_csv(
    "Grid_Stability_Training.csv",
    index=False
)

test_data.to_csv(
    "Grid_Stability_Test.csv",
    index=False
)

print("\nTraining and Test files saved successfully!")

# SELECT THE 12 INPUT FEATURES

features = [
    "tau1",
    "tau2",
    "tau3",
    "tau4",
    "p1",
    "p2",
    "p3",
    "p4",
    "g1",
    "g2",
    "g3",
    "g4"
]
# Training input
X_train = train_data[features]

# Training target
y_train = train_data["stabf"]

# Test input
X_test = test_data[features]

# Test target
y_test = test_data["stabf"]

print("\nNumber of input features:", len(features))
print("Training samples:", len(X_train))
print("Test samples:", len(X_test))

# ENCODE STABLE / UNSTABLE

encoder = LabelEncoder()

# Fit using training target
y_train = encoder.fit_transform(y_train)

# Use the same encoder for test target
y_test = encoder.transform(y_test)

print("\nTarget classes:")
print(encoder.classes_)

print("0 =", encoder.classes_[0])
print("1 =", encoder.classes_[1])

# STANDARDIZE THE DATA

# Scaling is required mainly for Logistic Regression and KNN.

scaler = StandardScaler()

# Fit scaler only using training data
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same scaler to test data
X_test_scaled = scaler.transform(X_test)

print("\nFeature scaling completed!")

# LOGISTIC REGRESSION

logistic_model = LogisticRegression(
    max_iter=1000
)

logistic_model.fit(
    X_train_scaled,
    y_train
)

logistic_prediction = logistic_model.predict(
    X_test_scaled
)

logistic_probability = logistic_model.predict_proba(
    X_test_scaled
)[:, 1]

print("\nLogistic Regression completed!")


# DECISION TREE

decision_tree = DecisionTreeClassifier(
    random_state=42
)

decision_tree.fit(
    X_train,
    y_train
)

tree_prediction = decision_tree.predict(
    X_test
)

tree_probability = decision_tree.predict_proba(
    X_test
)[:, 1]

print("Decision Tree completed!")


# K-NEAREST NEIGHBOUR

knn_model = KNeighborsClassifier(
    n_neighbors=5
)

knn_model.fit(
    X_train_scaled,
    y_train
)

knn_prediction = knn_model.predict(
    X_test_scaled
)

knn_probability = knn_model.predict_proba(
    X_test_scaled
)[:, 1]

print("KNN completed!")

# GAUSSIAN NAIVE BAYES

naive_bayes = GaussianNB()

naive_bayes.fit(
    X_train,
    y_train
)

nb_prediction = naive_bayes.predict(
    X_test
)

nb_probability = naive_bayes.predict_proba(
    X_test
)[:, 1]

print("Gaussian Naive Bayes completed!")


# RANDOM FOREST

random_forest = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

random_forest.fit(
    X_train,
    y_train
)

rf_prediction = random_forest.predict(
    X_test
)

rf_probability = random_forest.predict_proba(
    X_test
)[:, 1]

print("Random Forest completed!")

# CALCULATE RESULTS FOR ALL FIVE MODELS

# Logistic Regression

logistic_accuracy = accuracy_score(
    y_test,
    logistic_prediction
)

logistic_auc = roc_auc_score(
    y_test,
    logistic_probability
)

logistic_precision = precision_score(
    y_test,
    logistic_prediction,
    zero_division=0
)

logistic_recall = recall_score(
    y_test,
    logistic_prediction,
    zero_division=0
)

logistic_f1 = f1_score(
    y_test,
    logistic_prediction,
    zero_division=0
)

logistic_mcc = matthews_corrcoef(
    y_test,
    logistic_prediction
)


# Decision Tree

tree_accuracy = accuracy_score(
    y_test,
    tree_prediction
)

tree_auc = roc_auc_score(
    y_test,
    tree_probability
)

tree_precision = precision_score(
    y_test,
    tree_prediction,
    zero_division=0
)

tree_recall = recall_score(
    y_test,
    tree_prediction,
    zero_division=0
)

tree_f1 = f1_score(
    y_test,
    tree_prediction,
    zero_division=0
)

tree_mcc = matthews_corrcoef(
    y_test,
    tree_prediction
)


# KNN

knn_accuracy = accuracy_score(
    y_test,
    knn_prediction
)

knn_auc = roc_auc_score(
    y_test,
    knn_probability
)

knn_precision = precision_score(
    y_test,
    knn_prediction,
    zero_division=0
)

knn_recall = recall_score(
    y_test,
    knn_prediction,
    zero_division=0
)

knn_f1 = f1_score(
    y_test,
    knn_prediction,
    zero_division=0
)

knn_mcc = matthews_corrcoef(
    y_test,
    knn_prediction
)


# Gaussian Naive Bayes

nb_accuracy = accuracy_score(
    y_test,
    nb_prediction
)

nb_auc = roc_auc_score(
    y_test,
    nb_probability
)

nb_precision = precision_score(
    y_test,
    nb_prediction,
    zero_division=0
)

nb_recall = recall_score(
    y_test,
    nb_prediction,
    zero_division=0
)

nb_f1 = f1_score(
    y_test,
    nb_prediction,
    zero_division=0
)

nb_mcc = matthews_corrcoef(
    y_test,
    nb_prediction
)


# Random Forest

rf_accuracy = accuracy_score(
    y_test,
    rf_prediction
)

rf_auc = roc_auc_score(
    y_test,
    rf_probability
)

rf_precision = precision_score(
    y_test,
    rf_prediction,
    zero_division=0
)

rf_recall = recall_score(
    y_test,
    rf_prediction,
    zero_division=0
)

rf_f1 = f1_score(
    y_test,
    rf_prediction,
    zero_division=0
)

rf_mcc = matthews_corrcoef(
    y_test,
    rf_prediction
)

# CREATE FINAL COMPARISON TABLE

results = pd.DataFrame({

    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "K-Nearest Neighbour",
        "Gaussian Naive Bayes",
        "Random Forest"
    ],

    "Accuracy": [
        logistic_accuracy,
        tree_accuracy,
        knn_accuracy,
        nb_accuracy,
        rf_accuracy
    ],

    "AUC Score": [
        logistic_auc,
        tree_auc,
        knn_auc,
        nb_auc,
        rf_auc
    ],

    "Precision": [
        logistic_precision,
        tree_precision,
        knn_precision,
        nb_precision,
        rf_precision
    ],

    "Recall": [
        logistic_recall,
        tree_recall,
        knn_recall,
        nb_recall,
        rf_recall
    ],

    "F1 Score": [
        logistic_f1,
        tree_f1,
        knn_f1,
        nb_f1,
        rf_f1
    ],

    "MCC Score": [
        logistic_mcc,
        tree_mcc,
        knn_mcc,
        nb_mcc,
        rf_mcc
    ]
})

# Round values to four decimal places

results = results.round(4)

#  DISPLAY FINAL COMPARISON

print("\n")
print("=" * 90)
print("FINAL COMPARISON OF FIVE MACHINE LEARNING MODELS")
print("=" * 90)

display(results)



# FIND THE BEST MODEL

# F1 Score is used to identify the best overall model.

best_index = results["F1 Score"].idxmax()

best_model = results.loc[
    best_index,
    "Model"
]


# HIGHLIGHT THE BEST MODEL

def highlight_best(row):

    if row.name == best_index:
        return ["background-color: lightgreen"] * len(row)

    return [""] * len(row)


print("\nBest model is highlighted in green:")

display(
    results.style.apply(
        highlight_best,
        axis=1
    )
)


# DISPLAY BEST MODEL DETAILS
print("\n")
print("=" * 50)
print("BEST PERFORMING MODEL")
print("=" * 50)

print("Best Model :", best_model)
print("Accuracy   :", results.loc[best_index, "Accuracy"])
print("AUC Score  :", results.loc[best_index, "AUC Score"])
print("Precision  :", results.loc[best_index, "Precision"])
print("Recall     :", results.loc[best_index, "Recall"])
print("F1 Score   :", results.loc[best_index, "F1 Score"])
print("MCC Score  :", results.loc[best_index, "MCC Score"])


# SAVE FINAL RESULTS

results.to_csv(
    "Grid_Stability_ML_Results.csv",
    index=False
)

print("\nFinal results saved as:")
print("Grid_Stability_ML_Results.csv")


# TEST ONE NEW GRID SAMPLE

# Enter one new sample containing the 12 input features.

new_sample = pd.DataFrame([{

    "tau1": 2.0,
    "tau2": 1.0,
    "tau3": 0.0,
    "tau4": 0.5,

    "p1": 3.0,
    "p2": -1.0,
    "p3": -1.2,
    "p4": -0.8,

    "g1": 0.6,
    "g2": 0.7,
    "g3": 0.8,
    "g4": 0.9
}])


# Scale the sample for Logistic Regression and KNN

new_sample_scaled = scaler.transform(
    new_sample[features]
)


# Make predictions

new_predictions = {

    "Logistic Regression":
        logistic_model.predict(new_sample_scaled)[0],

    "Decision Tree":
        decision_tree.predict(new_sample[features])[0],

    "K-Nearest Neighbour":
        knn_model.predict(new_sample_scaled)[0],

    "Gaussian Naive Bayes":
        naive_bayes.predict(new_sample[features])[0],

    "Random Forest":
        random_forest.predict(new_sample[features])[0]
}

# Display predictions

print("\n")
print("=" * 50)
print("PREDICTION FOR ONE NEW GRID SAMPLE")
print("=" * 50)

for model_name, prediction in new_predictions.items():

    result = encoder.inverse_transform([prediction])[0]

    print(
        model_name,
        ":",
        result
    )
```
